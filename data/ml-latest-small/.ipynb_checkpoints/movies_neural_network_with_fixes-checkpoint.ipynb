# Movie Recommendation System with Neural Networks
## MovieLens Dataset - Content-Based + Collaborative Filtering

**Goal:** Build a neural network that recommends movies to users based on:
1. Movies they've rated highly (similar movie features)
2. What similar users liked (collaborative filtering)
3. Movie genres and content features

**Dataset:** MovieLens 100K - 100,000 ratings from 600 users on 9,000 movies

```python
import pandas as pd
import numpy as np

np.random.seed(42)
tf.random.set_seed(42)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MultiLabelBinarizer
from sklearn.metrics import mean_squared_error, mean_absolute_error
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
```

```python
# Load datasets
movies_df = pd.read_csv('../Desktop/movie-recommendation/ml-latest-small/movies.csv')
ratings_df = pd.read_csv('../Desktop/movie-recommendation/ml-latest-small/ratings.csv')
tags_df = pd.read_csv('../Desktop/movie-recommendation/ml-latest-small/tags.csv')
```

```python
print("Movies dataset:")
print(f"  Shape: {movies_df.shape}")
print(f"  Unique movies: {movies_df['movieId'].nunique()}")

print("\nRatings dataset:")
print(f"  Shape: {ratings_df.shape}")
print(f"  Unique users: {ratings_df['userId'].nunique()}")
print(f"  Unique movies rated: {ratings_df['movieId'].nunique()}")
print(f"  Rating range: {ratings_df['rating'].min()} - {ratings_df['rating'].max()}")

print("\nTags dataset:")
print(f"  Shape: {tags_df.shape}")
```

```python
# Process genres - split by pipe and create binary features
movies_df['genres_list'] = movies_df['genres'].str.split('|')

# Get all unique genres
all_genres = set()
for genres in movies_df['genres_list']:
    all_genres.update(genres)

all_genres = sorted(list(all_genres - {'(no genres listed)'}))
print(f"Unique genres: {len(all_genres)}")
print(all_genres)
```

```python
# Create binary genre features
for genre in all_genres:
    movies_df[f'genre_{genre}'] = movies_df['genres_list'].apply(lambda x: 1 if genre in x else 0)

# Get genre feature columns
genre_cols = [f'genre_{genre}' for genre in all_genres]

movies_df[['title', 'genres'] + genre_cols[:5]].head()
```

```python
movies_df['year'] = movies_df['title'].str.extract(r'\((\d{4})\)').astype(float)
movies_df['year'].isna().sum()
```

```python
movies_df['year'] = movies_df['year'].dropna()

# Normalize year to 0-1 range
movies_df['year_normalized'] = (movies_df['year'] - movies_df['year'].min()) / (movies_df['year'].max() - movies_df['year'].min())

print(f"Year range: {movies_df['year'].min():.0f} - {movies_df['year'].max():.0f}")
```

```python
# Calculate movie popularity (number of ratings)
movie_popularity = ratings_df.groupby('movieId').size().reset_index(name='num_ratings')
movies_df = movies_df.merge(movie_popularity, on='movieId', how='left')
movies_df['num_ratings'] = movies_df['num_ratings'].fillna(0)
```

```python
movies_df['popularity_normalized'] = np.log1p(movies_df['num_ratings']) / np.log1p(movies_df['num_ratings'].max())

# Calculate average rating
movie_avg_rating = ratings_df.groupby('movieId')['rating'].mean().reset_index(name='avg_rating')
movies_df = movies_df.merge(movie_avg_rating, on='movieId', how='left')
movies_df['avg_rating'] = movies_df['avg_rating'].fillna(3.0)  # Fill with neutral rating

# Normalize average rating
movies_df['avg_rating_normalized'] = (movies_df['avg_rating'] - 0.5) / 5.0

print("Movie features created:")
print(f"  Genre features: {len(genre_cols)}")
print(f"  Metadata features: year, popularity, avg_rating")
print(f"  Total features per movie: {len(genre_cols) + 3}")
```

```python
# Create final movie feature matrix
movie_feature_cols = genre_cols + ['year_normalized', 'popularity_normalized', 'avg_rating_normalized']
movie_features = movies_df[['movieId'] + movie_feature_cols].copy()

print(f"Movie features shape: {movie_features.shape}")
movie_features.head()
```

```python
# Create user and movie ID mappings
user_ids = data['userId'].unique()
movie_ids = data['movieId'].unique()

user_id_to_idx = {user_id: idx for idx, user_id in enumerate(user_ids)}
movie_id_to_idx = {movie_id: idx for idx, movie_id in enumerate(movie_ids)}

idx_to_user_id = {idx: user_id for user_id, idx in user_id_to_idx.items()}
idx_to_movie_id = {idx: movie_id for movie_id, idx in movie_id_to_idx.items()}

# Map IDs to indices
data['user_idx'] = data['userId'].map(user_id_to_idx)
data['movie_idx'] = data['movieId'].map(movie_id_to_idx)

print(f"\nNumber of unique users: {len(user_ids)}")
print(f"Number of unique movies: {len(movie_ids)}")
```

```python
# Prepare features and labels
X_user_idx = data['user_idx'].values
X_movie_idx = data['movie_idx'].values
X_movie_features = data[movie_feature_cols].values
y = data['rating'].values

print(f"User indices shape: {X_user_idx.shape}")
print(f"Movie indices shape: {X_movie_idx.shape}")
print(f"Movie features shape: {X_movie_features.shape}")
print(f"Ratings shape: {y.shape}")
```

```python
# Split data into train/validation/test sets
from sklearn.model_selection import train_test_split

# First split: 80% train+val, 20% test
indices = np.arange(len(y))
train_val_idx, test_idx = train_test_split(indices, test_size=0.2, random_state=42)

# Second split: 80% train, 20% val (of train+val)
train_idx, val_idx = train_test_split(train_val_idx, test_size=0.2, random_state=42)

# Split all features
X_user_train = X_user_idx[train_idx]
X_user_val = X_user_idx[val_idx]
X_user_test = X_user_idx[test_idx]

X_movie_train = X_movie_idx[train_idx]
X_movie_val = X_movie_idx[val_idx]
X_movie_test = X_movie_idx[test_idx]

X_features_train = X_movie_features[train_idx]
X_features_val = X_movie_features[val_idx]
X_features_test = X_movie_features[test_idx]

y_train = y[train_idx]
y_val = y[val_idx]
y_test = y[test_idx]

print(f"Train set: {len(y_train)} ratings")
print(f"Validation set: {len(y_val)} ratings")
print(f"Test set: {len(y_test)} ratings")
```

```python
def build_collaborative_model(n_users, n_movies, embedding_dim=50):
    """
    Neural Collaborative Filtering (NCF) model
    Similar to your matrix factorization but with deep neural network
    """
    
    # User input and embedding
    user_input = layers.Input(shape=(1,), name='user_input')
    user_embedding = layers.Embedding(
        n_users, 
        embedding_dim, 
        name='user_embedding',
        embeddings_regularizer=keras.regularizers.l2(1e-6)
    )(user_input)
    user_vec = layers.Flatten(name='user_flatten')(user_embedding)
    
    # Movie input and embedding
    movie_input = layers.Input(shape=(1,), name='movie_input')
    movie_embedding = layers.Embedding(
        n_movies, 
        embedding_dim, 
        name='movie_embedding',
        embeddings_regularizer=keras.regularizers.l2(1e-6)
    )(movie_input)
    movie_vec = layers.Flatten(name='movie_flatten')(movie_embedding)
    
    # Concatenate user and movie embeddings
    concat = layers.Concatenate(name='concat')([user_vec, movie_vec])
    
    # Deep neural network
    dense1 = layers.Dense(128, activation='relu', name='dense1')(concat)
    dropout1 = layers.Dropout(0.3)(dense1)
    
    dense2 = layers.Dense(64, activation='relu', name='dense2')(dropout1)
    dropout2 = layers.Dropout(0.3)(dense2)
    
    dense3 = layers.Dense(32, activation='relu', name='dense3')(dropout2)
    
    # Output layer - predict rating (0.5 to 5.0)
    output = layers.Dense(1, activation='linear', name='output')(dense3)
    
    model = Model(inputs=[user_input, movie_input], outputs=output, name='CollaborativeFiltering')
    
    return model

# Build the model
collab_model = build_collaborative_model(len(user_ids), len(movie_ids), embedding_dim=50)

# Compile
collab_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

collab_model.summary()
```

### Model 2: Hybrid Model (Embeddings + Content Features)

```python
def build_hybrid_model(n_users, n_movies, n_features, embedding_dim=50):
    """
    Hybrid recommendation model combining:
    - User/Movie embeddings (collaborative filtering)
    - Movie content features (content-based filtering)
    """
    
    # User embedding branch
    user_input = layers.Input(shape=(1,), name='user_input')
    user_embedding = layers.Embedding(
        n_users, 
        embedding_dim, 
        name='user_embedding',
        embeddings_regularizer=keras.regularizers.l2(1e-6)
    )(user_input)
    user_vec = layers.Flatten(name='user_flatten')(user_embedding)
    
    # Movie embedding branch
    movie_input = layers.Input(shape=(1,), name='movie_input')
    movie_embedding = layers.Embedding(
        n_movies, 
        embedding_dim, 
        name='movie_embedding',
        embeddings_regularizer=keras.regularizers.l2(1e-6)
    )(movie_input)
    movie_vec = layers.Flatten(name='movie_flatten')(movie_embedding)
    
    # Content features branch
    content_input = layers.Input(shape=(n_features,), name='content_input')
    content_dense1 = layers.Dense(64, activation='relu', name='content_dense1')(content_input)
    content_dense2 = layers.Dense(32, activation='relu', name='content_dense2')(content_dense1)
    
    # Combine all branches
    concat = layers.Concatenate(name='concat')([user_vec, movie_vec, content_dense2])
    
    # Deep layers
    dense1 = layers.Dense(128, activation='relu', name='dense1')(concat)
    dropout1 = layers.Dropout(0.3)(dense1)
    
    dense2 = layers.Dense(64, activation='relu', name='dense2')(dropout1)
    dropout2 = layers.Dropout(0.2)(dense2)
    
    dense3 = layers.Dense(32, activation='relu', name='dense3')(dropout2)
    
    # Output
    output = layers.Dense(1, activation='linear', name='output')(dense3)
    
    model = Model(
        inputs=[user_input, movie_input, content_input], 
        outputs=output, 
        name='HybridRecommender'
    )
    
    return model

# Build the hybrid model
hybrid_model = build_hybrid_model(len(user_ids), len(movie_ids), len(movie_feature_cols), embedding_dim=50)

# Compile
hybrid_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

hybrid_model.summary()
```

### Train Collaborative Filtering Model

```python
# Callbacks
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)

# Train
print("Training Collaborative Filtering Model...")
collab_history = collab_model.fit(
    [X_user_train, X_movie_train],
    y_train,
    validation_data=([X_user_val, X_movie_val], y_val),
    epochs=50,
    batch_size=256,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```

### Train Hybrid Model

```python
# Train hybrid model
print("Training Hybrid Model...")
hybrid_history = hybrid_model.fit(
    [X_user_train, X_movie_train, X_features_train],
    y_train,
    validation_data=([X_user_val, X_movie_val, X_features_val], y_val),
    epochs=50,
    batch_size=256,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```

```python
# Plot training history comparison
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Loss comparison
axes[0].plot(collab_history.history['loss'], label='Collab Train', alpha=0.7)
axes[0].plot(collab_history.history['val_loss'], label='Collab Val', alpha=0.7)
axes[0].plot(hybrid_history.history['loss'], label='Hybrid Train', alpha=0.7, linestyle='--')
axes[0].plot(hybrid_history.history['val_loss'], label='Hybrid Val', alpha=0.7, linestyle='--')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss (MSE)')
axes[0].set_title('Training Loss Comparison')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# MAE comparison
axes[1].plot(collab_history.history['mae'], label='Collab Train', alpha=0.7)
axes[1].plot(collab_history.history['val_mae'], label='Collab Val', alpha=0.7)
axes[1].plot(hybrid_history.history['mae'], label='Hybrid Train', alpha=0.7, linestyle='--')
axes[1].plot(hybrid_history.history['val_mae'], label='Hybrid Val', alpha=0.7, linestyle='--')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('MAE')
axes[1].set_title('Mean Absolute Error Comparison')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 7. Evaluate Models on Test Set

```python
# Evaluate Collaborative Filtering model
collab_pred = collab_model.predict([X_user_test, X_movie_test], verbose=0)
collab_mse = mean_squared_error(y_test, collab_pred)
collab_mae = mean_absolute_error(y_test, collab_pred)
collab_rmse = np.sqrt(collab_mse)

# Evaluate Hybrid model
hybrid_pred = hybrid_model.predict([X_user_test, X_movie_test, X_features_test], verbose=0)
hybrid_mse = mean_squared_error(y_test, hybrid_pred)
hybrid_mae = mean_absolute_error(y_test, hybrid_pred)
hybrid_rmse = np.sqrt(hybrid_mse)

# Display results
print("=" * 60)
print("TEST SET PERFORMANCE")
print("=" * 60)

print("\nCollaborative Filtering Model:")
print(f"  RMSE: {collab_rmse:.4f}")
print(f"  MAE:  {collab_mae:.4f}")
print(f"  MSE:  {collab_mse:.4f}")

print("\nHybrid Model (Embeddings + Content):")
print(f"  RMSE: {hybrid_rmse:.4f}")
print(f"  MAE:  {hybrid_mae:.4f}")
print(f"  MSE:  {hybrid_mse:.4f}")

print("\nImprovement:")
improvement = ((collab_rmse - hybrid_rmse) / collab_rmse) * 100
print(f"  Hybrid model is {abs(improvement):.2f}% {'better' if improvement > 0 else 'worse'} than Collaborative model")

print("\nNote: Lower values are better. RMSE/MAE show average prediction error in rating units (0.5-5.0 scale)")
```

## 8. Generate Movie Recommendations

Now let's use our trained models to recommend movies to users!

```python
def recommend_movies_for_user(user_id, model, model_type='collaborative', top_n=10):
    """
    Recommend top N movies for a given user
    
    Args:
        user_id: Original user ID from the dataset
        model: Trained recommendation model
        model_type: 'collaborative' or 'hybrid'
        top_n: Number of recommendations to return
    """
    
    # Check if user exists
    if user_id not in user_id_to_idx:
        print(f"User {user_id} not found in dataset!")
        return None
    
    user_idx = user_id_to_idx[user_id]
    
    # Get movies this user has already rated
    rated_movies = set(ratings_df[ratings_df['userId'] == user_id]['movieId'].values)
    
    # Get all movies the user hasn't rated
    all_movie_ids = movie_features['movieId'].values
    unrated_movies = [mid for mid in all_movie_ids if mid not in rated_movies and mid in movie_id_to_idx]
    
    if len(unrated_movies) == 0:
        print(f"User {user_id} has rated all movies!")
        return None
    
    # Prepare inputs for prediction
    user_indices = np.array([user_idx] * len(unrated_movies))
    movie_indices = np.array([movie_id_to_idx[mid] for mid in unrated_movies])
    
    # Predict ratings
    if model_type == 'collaborative':
        predictions = model.predict([user_indices, movie_indices], verbose=0).flatten()
    else:  # hybrid
        # Get movie features for unrated movies
        movie_feats = movie_features[movie_features['movieId'].isin(unrated_movies)][movie_feature_cols].values
        predictions = model.predict([user_indices, movie_indices, movie_feats], verbose=0).flatten()
    
    # Create recommendations dataframe
    recommendations = pd.DataFrame({
        'movieId': unrated_movies,
        'predicted_rating': predictions
    })
    
    # Sort by predicted rating
    recommendations = recommendations.sort_values('predicted_rating', ascending=False)
    
    # Add movie details
    recommendations = recommendations.merge(movies_df[['movieId', 'title', 'genres', 'year']], on='movieId')
    
    return recommendations.head(top_n)


def show_user_history(user_id, top_n=10):
    """
    Show what movies a user has rated highly in the past
    """
    user_ratings = ratings_df[ratings_df['userId'] == user_id].copy()
    user_ratings = user_ratings.merge(movies_df[['movieId', 'title', 'genres', 'year']], on='movieId')
    user_ratings = user_ratings.sort_values('rating', ascending=False)
    
    return user_ratings[['title', 'genres', 'year', 'rating']].head(top_n)
```

```python
# Example: Generate recommendations for a random user
sample_user_id = np.random.choice(user_ids)

print("=" * 80)
print(f"MOVIE RECOMMENDATIONS FOR USER {sample_user_id}")
print("=" * 80)

# Show what this user liked in the past
print("\n📽️  MOVIES THIS USER RATED HIGHLY:")
print("=" * 80)
user_history = show_user_history(sample_user_id, top_n=5)
for idx, row in user_history.iterrows():
    print(f"⭐ {row['rating']:.1f}/5.0 - {row['title']}")
    print(f"         Genres: {row['genres']}\n")
```

```python
# Get recommendations from Hybrid model
print("\n🎬 TOP 10 RECOMMENDATIONS (Hybrid Model):")
print("=" * 80)
hybrid_recs = recommend_movies_for_user(sample_user_id, hybrid_model, model_type='hybrid', top_n=10)

for idx, row in hybrid_recs.iterrows():
    print(f"{row.name + 1}. {row['title']}")
    print(f"   Predicted Rating: ⭐ {row['predicted_rating']:.2f}/5.0")
    print(f"   Genres: {row['genres']}")
    print()
```

## New User Recommendations Using Neural Network

Get movie recommendations for new users based on their genre preferences using the trained neural network.

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_for_new_user(user_genres, model, movies_df, genre_cols, 
                          movie_feature_cols, movie_id_to_idx, top_n=10):
    
    available_genres = [col.replace('genre_', '') for col in genre_cols]
    
    genre_mask = pd.Series([False] * len(movies_df), index=movies_df.index)
    for genre in user_genres:
        col = f'genre_{genre}'
        if col in movies_df.columns:
            genre_mask |= (movies_df[col] == 1)
    
    candidates = movies_df[genre_mask].copy()
    candidates = candidates[candidates['movieId'].isin(movie_id_to_idx.keys())]
    
    if len(candidates) == 0:
        return None
    
    user_indices = np.zeros(len(candidates), dtype=int)
    movie_indices = np.array([movie_id_to_idx[mid] for mid in candidates['movieId'].values])
    movie_feats = candidates[movie_feature_cols].values
    
    predicted_ratings = model.predict(
        [user_indices, movie_indices, movie_feats],
        verbose=0
    ).flatten()
    
    user_profile = np.zeros(len(available_genres))
    for i, genre in enumerate(available_genres):
        if genre in user_genres:
            user_profile[i] = 1
    
    movie_genres = candidates[genre_cols].values
    similarities = cosine_similarity([user_profile], movie_genres)[0]
    
    candidates['predicted_rating'] = predicted_ratings
    candidates['genre_match'] = similarities
    candidates['final_score'] = predicted_ratings * 0.7 + similarities * 5 * 0.3
    
    recommendations = candidates.sort_values('final_score', ascending=False)
    
    return recommendations[['title', 'genres', 'year', 'avg_rating', 'num_ratings',
                           'predicted_rating', 'genre_match', 'final_score']].head(top_n)
```

### Example: Get Recommendations

```python
MY_GENRES = ['Sci-Fi', 'Horror']

recommendations = recommend_for_new_user(
    user_genres=MY_GENRES,
    model=hybrid_model,
    movies_df=movies_df,
    genre_cols=genre_cols,
    movie_feature_cols=movie_feature_cols,
    movie_id_to_idx=movie_id_to_idx,
    top_n=10
)

print(f"Recommendations based on: {', '.join(MY_GENRES)}")
print("="*80)

for idx, (i, movie) in enumerate(recommendations.iterrows(), 1):
    print(f"{idx}. {movie['title']} ({int(movie['year'])})")
    print(f"   Predicted Rating: {movie['predicted_rating']:.2f}/5.0")
    print(f"   Actual Avg Rating: {movie['avg_rating']:.1f}/5.0 ({int(movie['num_ratings'])} ratings)")
    print(f"   Genre Match: {movie['genre_match']*100:.0f}%")
    print(f"   Genres: {movie['genres']}")
    print()
```

```python
# ADD: user-aware split (prevents validation leakage)
from sklearn.model_selection import train_test_split

train_users, test_users = train_test_split(
    ratings.userId.unique(), test_size=0.2, random_state=42
)

train_df = ratings[ratings.userId.isin(train_users)]
test_df  = ratings[ratings.userId.isin(test_users)]

train_df, val_df = train_test_split(train_df, test_size=0.2, random_state=42)
```

```python
# ADD: regularized model (same idea, safer training)
from tensorflow.keras.layers import Input, Embedding, Dense, Concatenate, Flatten, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping

user_input = Input(shape=(1,))
movie_input = Input(shape=(1,))
genre_input = Input(shape=(X_movie_features.shape[1],))
user_feat_input = Input(shape=(3,))

user_emb = Embedding(num_users, 32, embeddings_regularizer=l2(1e-6))(user_input)
movie_emb = Embedding(num_movies, 32, embeddings_regularizer=l2(1e-6))(movie_input)

x = Concatenate()([
    Flatten()(user_emb),
    Flatten()(movie_emb),
    genre_input,
    user_feat_input
])

x = Dense(128, activation='relu')(x)
x = Dropout(0.4)(x)
x = Dense(64, activation='relu')(x)
output = Dense(1)(x)

model = Model([user_input, movie_input, genre_input, user_feat_input], output)

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss='mse'
)

model.summary()
```

```python
# ADD: stable training
early_stop = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

history = model.fit(
    [
        train_df.user_idx.values,
        train_df.movie_idx.values,
        X_movie_features[train_df.index],
        train_df[['age_group','likes_scifi','likes_drama']].values
    ],
    train_df.rating.values,
    validation_data=(
        [
            val_df.user_idx.values,
            val_df.movie_idx.values,
            X_movie_features[val_df.index],
            val_df[['age_group','likes_scifi','likes_drama']].values
        ],
        val_df.rating.values
    ),
    epochs=20,
    batch_size=256,
    callbacks=[early_stop]
)
```
